The strategy involves using a Large Language Model to evaluate the attractiveness of each stock in the Russell 1000 index daily. The model autonomously searches the web, filters sources, and synthesizes information into quantitative predictions. The strategy focuses on longing the top 20 highest-ranked stocks based on the model's predictions.

## Phase 1 — Trading Context & Objectives

### Configuration
In this phase, we define the trading universe, parameters, and hypothesis for the strategy.

```python
# Configuration
UNIVERSE = ['AAPL', 'MSFT']
TOP_N = 20
HYPOTHESIS = "Longing the top 20 highest-ranked stocks based on AI predictions will generate positive alpha."
```

## Phase 2 — Data Download & Feature Computation

### Data Download
In this phase, we download market data for the specified universe using yfinance.

```python
import yfinance as yf
import pandas as pd
import numpy as np

# Download data
data = yf.download(UNIVERSE, start='2020-01-01', end='2023-01-01', group_by='ticker')['Adj Close']
```

## Phase 3 — Signal Generation & Portfolio Construction

### Signal Generation
In this phase, we generate signals based on the AI predictions.

```python
# Placeholder for AI predictions
ai_predictions = pd.DataFrame(np.random.rand(len(data), len(UNIVERSE)), index=data.index, columns=UNIVERSE)

# Rank stocks based on AI predictions
ai_predictions_ranked = ai_predictions.rank(axis=1, ascending=False)

# Generate signals
signals = (ai_predictions_ranked <= TOP_N).astype(int) - 0.5
```

## Phase 4 — Vectorized Backtest

### Backtest
In this phase, we perform a vectorized backtest of the strategy.

```python
# Shift signals forward by 1 period to avoid look-ahead bias
signals_shifted = signals.shift(1)

# Calculate portfolio returns
portfolio_returns = (signals_shifted * data.pct_change()).sum(axis=1)
```

## Phase 5 — Performance Metrics

### Performance Evaluation
In this phase, we evaluate the performance of the strategy using various metrics.

```python
import matplotlib.pyplot as plt
from scipy.stats import sharpe

# Calculate performance metrics
sharpe_ratio = sharpe(portfolio_returns.dropna().mean(), portfolio_returns.dropna().std())
max_drawdown = (portfolio_returns.cumsum().expanding().max() - portfolio_returns.cumsum()).max()

# Plot equity curve
plt.plot(portfolio_returns.cumsum())
plt.title('Equity Curve')
plt.show()

print(f'Sharpe Ratio: {sharpe_ratio}')
print(f'Max Drawdown: {max_drawdown}')
```

## Phase 6 — Monitoring Stub

### Monitoring
In this phase, we create a function to print daily P&L and current positions given live data.

```python
def monitor_portfolio(signals, prices):
    positions = signals.iloc[-1]
    daily_pnl = (positions * prices.pct_change().iloc[-1]).sum()
    print(f'Daily P&L: {daily_pnl}')
    print(f'Current Positions: {positions}')

# Example usage
monitor_portfolio(signals, data)
```
